**Integration Into the Main Dataset**

Two new categorical variables, region and sub-region, were added to the life_exp_clean_2 DataFrame using a multi-loop classification system:

- A dictionary (sub_regions_dict) was created to map sub-regions to their corresponding regions.
- An iterative loop assigned each country to its respective region and sub-region.
- The classification was then stored in lists and appended to the main dataset.

<div style="page-break-after: always;"></div>

The following code snippet illustrates the classification process:
```python
# Generate lists of sub-regions
sub_region_keys = list(countries_by_region.keys())
sub_regions_dict = {
    'Asian region': sub_region_keys[0:5],
    'European region': sub_region_keys[5:9],
    'African region': sub_region_keys[9:14],
    'American region': sub_region_keys[14:18],
    'Oceania region': sub_region_keys[18:22]
}

# Assign region and sub-region classifications
region_classification_list = []
sub_region_classification_list = []
for country in list(life_exp_clean_2['country']):
    for sub_region in countries_by_region:
        if country in countries_by_region[sub_region]:
            for regions in sub_regions_dict:
                if sub_region in sub_regions_dict[regions]:
                    region_classification_list.append(regions)
                    sub_region_classification_list.append(sub_region)

# Add region and sub-region classifications as string-type columns
life_exp_clean_2['region'] = region_classification_list.astype('string')
life_exp_clean_2['sub_region'] = sub_region_classification_list.astype('string')
```
<br>
The inclusion of regional and sub-regional classifications allows for structured comparative analysis across different geographic areas.

### **2. Assigning Income Classes Based on World Bank Classifications**

To further enrich the dataset, each country was assigned an income class based on the World Bank’s classification thresholds. This classification aids in examining how economic factors impact life expectancy and related health indicators.

**Income Classification Thresholds**

The following World Bank income thresholds (2015 values) were applied:

- Low Income: GNI per capita ≤ 1,045 USD
- Lower-Middle Income: GNI per capita ≤ 4,125 USD
- Upper-Middle Income: GNI per capita ≤ 12,735 USD
- High Income: GNI per capita > 12,735 USD

<div style="page-break-after: always;"></div>

**Automated Income Classification Process**

To simplify income classification of countries for this project, countries were assigned based on 2015 income classes. To categorize each country, the following approach was implemented:

1. The income class was assigned based on each country’s Gross National Income (GNI) per capita in 2015.
2. The classification was determined using a lambda function within an apply() method.
3. Missing values were forward-filled (ffill()) to ensure consistency.

The following Python code was used:

```python
life_exp_clean_2['income_class'] = life_exp_clean_2['gni'][life_exp_clean_2['year'] == 2015].apply(
    lambda x: 'Low income' if x <= 1045 else
              'Lower-middle income' if x <= 4125 else
              'Upper-middle income' if x <= 12735 else
              'High income'
)

# Forward-fill missing values
life_exp_clean_2['income_class'] = life_exp_clean_2['income_class'].ffill()
```
<br>
This process ensures that all 176 countries are correctly classified according to regional, sub-regional, and income groupings, enriching the dataset for advanced analysis and visualization.

### 2.4.2 Creating Multiple Imputation Class

Missing data poses a significant challenge in any analysis, particularly when working with economic and health indicators where data collection may not be consistently available across all regions and years. To address the missing values in the health_expenditure_percentage and total_expenditure variables, a Multiple Imputation Class was developed. This class leverages an advanced imputation technique by identifying countries with similar economic and health characteristics to serve as reference data for missing entries.

**Identifying Missing Data Patterns**

A detailed analysis of missing data patterns revealed that for health_expenditure_percentage and total_expenditure, five countries—Afghanistan, Iraq, Montenegro, Timor-Leste, and Zimbabwe—had missing data primarily in the earlier years (2000 onward). Given that data collection improved in later years, it was inferred that the missing data was likely missing at random (MAR) rather than completely missing without a pattern. This assumption supported the approach of using similar countries with complete data for imputation.

**Exploring Different Imputation Approaches**

Initially, multiple strategies were tested to train the IterativeImputer:

1. Using regional data (aggregating data within broader regions).
2. Using sub-regional data (a more localized approach).
3. Using the entire dataset for training the imputer.

**None of these methods** produced reliable results due to the limited number of observations per country (only 16 yearly data points). This limitation meant that a single fitted value was often used across all missing entries, reducing the accuracy and robustness of the imputation.

<div style="page-break-after: always;"></div>

**Final Imputation Strategy**

The optimal method involved finding countries with similar characteristics (excluding the missing variable) and then training the imputer on these selected countries. This approach allowed for a more contextually relevant imputation by ensuring that missing values were filled using countries with comparable economic and health profiles.

The imputation class, CountryImputer, follows these steps:

1. Identify similar countries based on shared economic, mortality and health characteristics.
2. Train an imputer using data from these selected countries, excluding the variable being imputed.
3. Apply imputation to fill missing values for the target country.

This methodology accounted for structural similarities between countries rather than relying on arbitrary regional groupings.

**Class Implementation and Functionality**

**Step 1:** Finding Similar Countries

The find_similar_countries method identifies relevant reference countries using a tolerance-based filtering approach:

- A set of core variables (life_expectancy, adult_mortality, infant_deaths, under_five_deaths, bmi_male, bmi_female, gdp_per_capita, and gni) are considered.
- The method calculates maximum and minimum values for each variable within the target country.
- A tolerance margin is applied to these values to create a range within which other countries should fall.
- Countries whose variable values fall within this tolerance range are selected as reference data.

This method ensures that only countries with comparable economic and health characteristics are used in the imputation process.

**Step 2:** Imputing Missing Data

The impute_country method applies Iterative Imputation to the selected reference data:

- The data is first normalized (e.g., GDP per capita and life expectancy are scaled for consistency).
- Certain variables (e.g., adult_mortality, infant_deaths) are inverted since lower values correlate with higher economic and health indicators.
- The IterativeImputer is trained using the selected reference countries, excluding the missing variable.
- The trained imputer is used to predict missing values for the target country.

<div style="page-break-after: always;"></div>

**Addressing Economic Fluctuations**

One of the key challenges encountered was the economic fluctuations over time, which could not always be accounted for with the limited available data. Since imputation relies on patterns within existing data, unpredictable shifts in economic trends (such as political instability or sudden growth spurts) could reduce accuracy. Despite this limitation, the chosen method produced more reliable estimates than other tested approaches.

**Example Usage of the Imputation Class**

To demonstrate the application of the CountryImputer class, consider the following example where Afghanistan's health_expenditure_percentage is imputed:
```python
finder = CountryImputer(life_exp_clean_2, 'Afghanistan', 'health_expenditure_percentage')
considered_countries = finder.find_similar_countries(10, ['bmi_male', 'total_expenditure', 'gni'])
hep_multiple_imp = finder.impute_country(considered_countries)
```
In this case:

- A 10% tolerance was applied to find similar countries.
- Certain variables (bmi_male, total_expenditure, gni) were excluded to focus on economic and mortality-related indicators.
- The resulting imputation was applied only to Afghanistan for the missing years.

**Outputs and Verification**

Throughout the imputation process, several diagnostic outputs were generated to ensure transparency:

1. Max and Min values per variable for the target country.
2. Max and Min values with applied tolerance to identify similar countries.
3. List of selected reference countries used for imputation.
4. Years with missing data to track imputed entries.
5. Final imputed dataset preview to validate the results.

**Key notes**

The development of this Multiple Imputation Class provides a structured and adaptable approach to handling missing data in economic and health datasets. By leveraging similarities between countries, this method ensures that missing values are imputed in a contextually relevant manner, improving the integrity and completeness of the dataset. While economic fluctuations remain a challenge, the approach provides a significantly better alternative to traditional imputation methods, ensuring that meaningful analyses can be conducted without the bias introduced by arbitrary data removal or simplistic imputation techniques.

<div style="page-break-after: always;"></div>

- **Standard Deviation:**
The standard deviation, which quantifies variation or dispersion in life expectancy for each year, was calculated with `np.std()`.<br>
Example:<br>
    ```python
    std = np.std((life_expectancy['life_expectancy'].where(life_expectancy.year == year)).dropna())
    ```
    <br>

- **Inter-quartile Range (IQR):**
The IQR was computed as the difference between the 75th percentile (Q3) and 25th percentile (Q1), providing a robust measure of spread.<br>
Example:<br>
    ```python
    q3, q1 = np.percentile(life_expectancy['life_expectancy'][life_expectancy['year'] == year], [75, 25])
    iqr = q3 - q1
    ```

After computing each metric, the resulting dictionaries were converted into DataFrames and merged into a final summary dataset:<br>
Example:<br>
```python
descriptive_stats_life_exp = le_all_means_df.merge(le_all_max_df, on='year')\
    .merge(le_all_min_df, on='year')\
    .merge(le_all_median_df, on='year')\
    .merge(le_all_std_df, on='year')\
    .merge(le_all_iqr_df, on='year')
```

### **2. Distribution Diagrams**

#### **How to read Distribution Diagrams**
A distribution diagram (such as a histogram) visualizes how a variable's values are spread across observations. In this project, distribution diagrams are used to show how life expectancy and GDP per capita vary across countries, years, regions, income classes, or GDP rankings.

**How to read diagram**
- **X-Axis:** Represents the range of values for the variable being analyzed (e.g., life expectancy in years or GDP per capita in USD).
- **Y-Axis:** Represents the frequency (for histograms) or probability density (for KDE plots) — i.e., how often values fall into a given range.
- **Shape:** The shape of the curve or bars indicates the pattern of distribution — whether it is normal, skewed, bimodal, or uniform.<br>

**What to look for**
- **Center:** Where most data points cluster — indicates the typical value.
- **Spread:** How wide the distribution is — reveals variability or consistency.
- **Skewness:**
    - **Right-skewed:** Tail on the right; common in GDP distributions due to very high-income outliers.
    - **Left-skewed:** Tail on the left; less common in life expectancy but can occur in poorer or conflict-affected countries.
- **Peaks (Modes):** One or more visible peaks can indicate sub-groups or regional/income clusters.
- **Outliers or Long Tails:** Extreme values that may impact interpretation or indicate unique conditions in certain countries.

These diagrams are especially helpful in identifying inequality, disparities, and shifts in distribution over time or across groups — all of which support the report’s core hypothesis on the relationship between GDP and life expectancy.

### **4. Boxplots**

Understanding Boxplots in This Report
Boxplots are powerful tools for comparing the distribution of a numerical variable across categories — in this case, life expectancy across regions, income classes, or years. They provide a concise summary of the data's spread, central tendency, and potential outliers.

#### **How to Read Boxplots**

**Reading boxplots**<br>
- **Box:**<br>
The box itself represents the interquartile range (IQR) — the middle 50% of values, from the 25th percentile (Q1) to the 75th percentile (Q3).
- **Line Inside the Box:**<br>
This line marks the median (Q2) — the middle value of the dataset.
- **Whiskers:**<br>
Lines extending from the box to the minimum and maximum values within 1.5×IQR from the lower and upper quartiles. They represent the bulk of the non-outlier data.
- **Dots Beyond Whiskers:**<br>
These are outliers, which fall outside the whisker range and may represent exceptional cases (e.g., extremely high or low life expectancy countries).

**What to look for**<br>
- **Median Line Position:**<br>
A higher median means better outcomes (e.g., higher life expectancy).
- **Box Height (IQR):**<br>
A tall box indicates greater variability; a short box shows consistency within the group.
- **Outliers:**<br>
Outliers can highlight countries with unique health or economic conditions, or may suggest data issues.
- **Comparisons Between Groups:**<br>
    - Are some groups consistently higher than others (e.g., high-income vs. low-income)?
    - Is the distribution tighter (less spread) for some groups, indicating uniform development?

### **5. Scatterplots**

**Understanding Scatterplots in This Report**

Scatterplots are used to visually explore the relationship between two continuous variables — in this case, typically life expectancy and GDP per capita. Each point on the graph represents a single country in a specific year, allowing us to analyze trends, correlations, and anomalies across the dataset.

#### **How to read Scatterplot**

**Reading scatterplot**
- **X-Axis:** Represents GDP per capita (USD) — the economic measure of a country’s wealth.
- **Y-Axis:** Represents life expectancy (in years) — the average number of years a person is expected to live.
- **Each Point:** A single observation — one country in one year.

Color or Shape (if used): May represent groupings such as region or income class to help distinguish patterns.

**What to look for**
- **General Pattern (Trend):**<br>
An upward trend (points rising as you move right) suggests a positive correlation — as GDP increases, so does life expectancy.
- **Strength of Relationship:**<br>
If points are tightly clustered along a line, the relationship is strong. If they are widely spread, the relationship is weaker.
- **Clusters:**<br>
Groupings of points may indicate similar characteristics within regions or income classes (e.g., low-income countries clustering at low GDP and life expectancy).
- **Outliers:**<br>
Points that lie far from the main trend may represent anomalies, conflict-affected countries, or nations with unique health systems (e.g., Cuba or the U.S.).
- **Flattening Curve at Higher GDP:**
A common pattern is diminishing returns — life expectancy increases rapidly with GDP at low-income levels but levels off among high-income countries.

These scatterplots are crucial in this project to visually test the hypothesis that higher economic prosperity leads to longer lives. They complement the statistical analysis and help illustrate exceptions, disparities, and structural patterns in global development.

### **6. Heatmaps and Correlation Matrices**

#### How to Read Heatmaps and Correlation Matrices

**Understanding Heatmaps and Correlation Matrices in This Report**<br>
Heatmaps and correlation matrices are used in this project to explore the strength and direction of linear relationships between numerical variables, such as life expectancy, GDP per capita, health expenditure, adult mortality, and more. These visuals help identify which factors are most strongly associated with life expectancy and economic indicators, guiding deeper statistical analysis.

**Reading Correlation Matrix (Displayed as a Heatmap)**
- **Axes (Rows and Columns):**<br>
Each axis represents the same set of variables — for example, life expectancy, GDP per capita, health expenditure (%), etc.
- **Cell Values:**<br>
Each cell shows the **correlation coefficient** (r) between the variable on the x-axis and the one on the y-axis:
    - +1 indicates a **perfect positive correlation**
    - -1 indicates a **perfect negative correlation**
    - 0 indicates no **linear relationship**
- **Color Intensity:**<br>
The color gradient (e.g., from blue to red) reflects the strength and direction of the correlation:
    - Darker/more intense colors = stronger correlation
    - Warm tones (e.g., red/orange) = positive correlation
    - Cool tones (e.g., blue/purple) = negative correlation

**What to look for**
- **Strong Positive Correlations:**<br>
High GDP per capita and life expectancy typically show strong positive correlations, often appearing as dark warm-colored cells (close to +1).
- **Strong Negative Correlations:**<br>
Variables like adult mortality or under-5 deaths usually have strong negative correlations with life expectancy (shown in cool colors).
- **Clusters of Related Variables:**<br>
Groups of variables that are strongly correlated among themselves may reflect related health or economic factors (e.g., GDP and health expenditure).
- **Comparison Across Subgroups:**<br>
When heatmaps are grouped by **region** or **income class**, differences in correlation patterns may reveal structural or developmental disparities.

These visuals are especially powerful for identifying **key predictors** of life expectancy and for validating whether hypothesized relationships hold consistently across different contexts.

# 9. Terminology (Glossary of Key Terms)

**Data Operations & Programming Concepts**
- **Data cleaning:** Removing or fixing incorrect, missing, or inconsistent data.
- **Data transformation:** Changing data format, structure, or values for analysis.
- **Function:** Reusable block of code that performs a specific task.
- **Method:** A function associated with an object (e.g., .dropna() in Pandas).
- **Variable:** A name that stores data in a program.
- **Class:** A blueprint for creating objects with attributes and behaviors.
- **Lambda function:** A small anonymous function used for quick operations.
- **List comprehension:** A concise way to create lists using loops and conditions.
- **Dictionary:** A data structure with key-value pairs.
- **Keys:** Unique identifiers in a dictionary.
- **Values:** Data stored in a dictionary, accessed via keys.
- **CSV files:** Text files that store tabular data, separated by commas.
- **NaN values:** Placeholder for missing or undefined data in a dataset.
- **Missing values:** Data entries that are empty or unavailable.
- **Imputer:** A tool or algorithm used to estimate and fill missing values in data.
- **CountryImputer:** A custom-built class for imputing missing values using data from similar countries.
- **make_pipeline:** A function from sklearn.pipeline to chain preprocessing and modeling steps.
- **PolynomialFeatures:** Transforms input features into polynomial terms.
- **np.mean():** NumPy function that calculates the average of values.
- **np.min()** / **np.max():** NumPy functions to find the smallest/largest values.
- **np.std():** NumPy function to compute standard deviation.
- **np.percentile():** Finds a percentile value in a dataset (e.g., 25th, 75th).
- **ffill():** Pandas method for forward-filling missing values.
- **astype():** Pandas method to change data type of a column.
- **merge():** Combines two DataFrames based on a common key.
- **reset_index():** Resets the index of a DataFrame.
- **dropna():** Removes rows with missing values.
- **groupby():** Aggregates data based on one or more categories.
- **apply():** Applies a function across a DataFrame axis.
- **nlargest()** / **nsmallest():** Returns rows with the top or bottom values.
- **tolerance:** A value range used to select similar data.
- **scaling:** Adjusting data values to a common scale for analysis.
- **normalization:** Rescaling data to have specific statistical properties (e.g., 0 to 1).

**Data Merging, Types & Handling**
- **Left join:** Keeps all rows from the left table and matches from the right.
- **Concatenation:** Combining dataframes or arrays side by side or top to bottom.
- **Categorical attributes**: Non-numeric values representing categories or groups.
- **Categorical operation:** Analysis or transformation applied to categorical data.
- **Dtype:** Data type of a variable (e.g., int, float, object).
- **int64:** Integer data type with 64-bit precision.
- **float64:** Floating point data type with 64-bit precision.
- **object:** General data type used for strings or mixed types in Pandas.
- **category:** A memory-efficient data type for storing categorical variables.
- **Memory:** The amount of RAM used to store data.

**Missing Data Methods**
- **Imputation:** Filling missing data using statistical or algorithmic methods.
- **Extrapolation:** Predicting future values using trends in available data.
- **Removal / Drop:** Excluding rows or columns with missing or invalid data.
- **IterativeImputer:** An imputer from Scikit-learn that models missing values using other features.
- **Training imputer:** The process of fitting the imputation model on known data.
- **Missing at Random (MAR):** Missing values that depend on observed data.
- **Missing Not at Random (MNAR):** Missing values that depend on unobserved data.

**Statistical Concepts**
- **Central tendency:** The center of a data distribution (e.g., mean, median).
- **Mean:** The average of a set of values.
- **Median:** The middle value in a sorted dataset.
- **Maximum:** The highest value in a dataset.
- **Minimum:** The lowest value in a dataset.
- **Standard deviation:** Measure of spread around the mean.
- **Inter-quartile range (IQR):** Difference between the 75th and 25th percentiles.
- **25th percentile:** Value below which 25% of data falls.
- **75th percentile:** Value below which 75% of data falls.
- **Spread:** The range or variability in a dataset.
- **Outliers:** Values that lie far outside the normal range of data.

- **Seaborn:** A Python data visualization library based on matplotlib.
- **Matplotlib:** Core Python library for creating static plots and graphs.

- **Numpy:** Python library for fast numerical computations, especially with arrays and matrices.
- **Scikit-learn:** A machine learning library for predictive modeling and preprocessing.
- **Heatmap and correlation matrix:** Visual tools showing variable relationships.
- **Scatterplot:** A graph showing the relationship between two numeric variables.
- **Boxplot:** Displays data spread, median, and outliers using a box and whiskers.
- **Distribution:** How data values are spread or grouped across a range.

**Analysis Techniques**
- **Correlation:** A measure of how two variables move together.
- **Polynomial regression:** A regression model that fits a curve to the data.
- **Linear regression:** A model that fits a straight line to predict a variable.
- **R² Score:** A metric that shows how well the model explains the data variance.
- **Mean Squared Error (MSE):** Average of squared prediction errors.
- **Aggregation:** Combining data values using functions like mean or sum.
- **Downstream analysis:** Subsequent steps in analysis using processed data.
