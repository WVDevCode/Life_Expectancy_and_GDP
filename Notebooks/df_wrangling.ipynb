```python
def create_png_dataframe(df, save_file, save_folder, fig_size):
    
    csv_buffer = StringIO()
    df.to_csv(csv_buffer, index=False)
    csv_buffer.seek(0)

    temp_df = pd.read_csv(csv_buffer)
    
    fig, ax = plt.subplots(figsize=fig_size)
    ax.set_frame_on(False)  # Remove frame
    ax.xaxis.set_visible(False)  # Remove x-axis
    ax.yaxis.set_visible(False)  # Remove y-axis

    os.makedirs(save_folder, exist_ok=True)
    save_path = os.path.join(save_folder, save_file)

    tbl = ax.table(
        cellText=temp_df.values,        # Only the data, no index.
        colLabels=temp_df.columns,        # Set column headers.
        cellLoc='center',
        loc='center',
        colWidths=[0.2] * len(temp_df.columns)
    )
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()
    plt.clf

# All countries
create_png_dataframe(descriptive_stats_life_exp, "Descriptive_stats-Life_Expectancy_All Data.png", "../Visualizations/Dataframes/Life_Expectancy", (6, 2))

# Africa
create_png_dataframe(descriptive_stats_life_exp_africa, "Descriptive_stats-Life_Expectancy_Africa.png", "../Visualizations/Dataframes/Life_Expectancy", (6, 2))

# Asia
create_png_dataframe(descriptive_stats_life_exp_asia, "Descriptive_stats-Life_Expectancy_Asia.png", "../Visualizations/Dataframes/Life_Expectancy", (6, 2))

# America
create_png_dataframe(descriptive_stats_life_exp_america, "Descriptive_stats-Life_Expectancy_America.png", "../Visualizations/Dataframes/Life_Expectancy", (6, 2))

# Europe
create_png_dataframe(descriptive_stats_life_exp_europe, "Descriptive_stats-Life_Expectancy_Europe.png", "../Visualizations/Dataframes/Life_Expectancy", (8, 4))

# Oceania
create_png_dataframe(descriptive_stats_life_exp_oceania, "Descriptive_stats-Life_Expectancy_Oceania.png", "../Visualizations/Dataframes/Life_Expectancy", (8, 4))
```

### Saving dataframes as png files

```python
def create_png_dataframe(df, save_file, save_folder, fig_size):
    
    csv_buffer = StringIO()
    df.to_csv(csv_buffer, index=False)
    csv_buffer.seek(0)

    temp_df = pd.read_csv(csv_buffer)
    
    fig, ax = plt.subplots(figsize=fig_size)
    ax.set_frame_on(False)  # Remove frame
    ax.xaxis.set_visible(False)  # Remove x-axis
    ax.yaxis.set_visible(False)  # Remove y-axis

    os.makedirs(save_folder, exist_ok=True)
    save_path = os.path.join(save_folder, save_file)

    tbl = ax.table(
        cellText=temp_df.values,        # Only the data, no index.
        colLabels=temp_df.columns,        # Set column headers.
        cellLoc='center',
        loc='center',
        colWidths=[0.2] * len(temp_df.columns)
    )
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()
    plt.clf

# All countries
create_png_dataframe(descriptive_stats_gdp_per_capita, "Descriptive_stats-GDP_per_Capita_All Data.png", "../Visualizations/Dataframes/GDP_per_Capita", (6, 2))

# Africa
create_png_dataframe(descriptive_stats_gdp_per_capita_africa, "Descriptive_stats-GDP_per_Capita_Africa.png", "../Visualizations/Dataframes/GDP_per_Capita", (6, 2))

# Asia
create_png_dataframe(descriptive_stats_gdp_per_capita_asia, "Descriptive_stats-GDP_per_Capita_Asia.png", "../Visualizations/Dataframes/GDP_per_Capita", (6, 2))

# America
create_png_dataframe(descriptive_stats_gdp_per_capita_america, "Descriptive_stats-GDP_per_Capita_America.png", "../Visualizations/Dataframes/GDP_per_Capita", (10, 2))

# Europe
create_png_dataframe(descriptive_stats_gdp_per_capita_europe, "Descriptive_stats-GDP_per_Capita_Europe.png", "../Visualizations/Dataframes/GDP_per_Capita", (6, 2))

# Oceania
create_png_dataframe(descriptive_stats_gdp_per_capita_oceania, "Descriptive_stats-GDP_per_Capita_Oceania.png", "../Visualizations/Dataframes/GDP_per_Capita", (6, 2))
```

Descriptive statistics on the relationship between life expectancy and gdp per capita to illustrate how the wealth of a country influences the quality of health of the country's population.

#### Create datasets for top and bottom 20 gdp per capita

##### Top 20 highest grossing per capita countries dataset

```python
# Manipulating Life Expectancy df descending to retrieve top 20 gdp per capita countries data
top_20_gdp_countries_list = life_expectancy.groupby('year').apply(lambda x: x.nlargest(20, 'gdp_per_capita'))['country'].unique()

le_top_20_gdp = life_expectancy[life_expectancy['country'].isin(top_20_gdp_countries_list)].reset_index(drop=True)

le_top_20_gdp.to_csv('../Datasets/Cleaned_datasets/EDA_Datasets/Top_Grossing_Countries-per_capita.csv')
```

##### 20 lowest grossing per capita countries dataset

```python
# Sorting Life Expectancy df ascending to retrieve bottom 20 gdp per capita countries data
le_bottom_20_gdp_countries_list = life_expectancy.groupby('year').apply(lambda x: x.nsmallest(20, 'gdp_per_capita'))['country'].unique()

le_bottom_20_gdp = life_expectancy[life_expectancy['country'].isin(le_bottom_20_gdp_countries_list)].reset_index(drop=True)

le_bottom_20_gdp.to_csv('../Datasets/Cleaned_datasets/EDA_Datasets/Lowest_Grossing_Countries-per_capita.csv')
```

```python
def create_png_dataframe(df, save_file, save_folder, fig_size):
    
    csv_buffer = StringIO()
    df.to_csv(csv_buffer, index=False)
    csv_buffer.seek(0)

    temp_df = pd.read_csv(csv_buffer)
    
    fig, ax = plt.subplots(figsize=fig_size)
    ax.set_frame_on(False)  # Remove frame
    ax.xaxis.set_visible(False)  # Remove x-axis
    ax.yaxis.set_visible(False)  # Remove y-axis

    os.makedirs(save_folder, exist_ok=True)
    save_path = os.path.join(save_folder, save_file)

    tbl = ax.table(
        cellText=temp_df.values,        # Only the data, no index.
        colLabels=temp_df.columns,        # Set column headers.
        cellLoc='center',
        loc='center',
        colWidths=[0.2] * len(temp_df.columns)
    )
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()
    plt.clf

# Top 20 GDP per Capita
create_png_dataframe(descriptive_stats_life_exp_top_gdp, "Descriptive_stats-Top_20_GDP.png", "../Visualizations/Dataframes/Top_and_Bottom_GDP", (6,2))

# Bottom 20 GDP per Capita
create_png_dataframe(descriptive_stats_life_exp_bottom_gdp, "Descriptive_stats-Bottom_20_GDP.png", "../Visualizations/Dataframes/Top_and_Bottom_GDP", (10,2))
```

```python
def create_png_dataframe(df, save_file, save_folder, fig_size):
    
    csv_buffer = StringIO()
    df.to_csv(csv_buffer, index=False)
    csv_buffer.seek(0)

    temp_df = pd.read_csv(csv_buffer)
    
    fig, ax = plt.subplots(figsize=fig_size)
    ax.set_frame_on(False)  # Remove frame
    ax.xaxis.set_visible(False)  # Remove x-axis
    ax.yaxis.set_visible(False)  # Remove y-axis

    os.makedirs(save_folder, exist_ok=True)
    save_path = os.path.join(save_folder, save_file)

    tbl = ax.table(
        cellText=temp_df.values,        # Only the data, no index.
        colLabels=temp_df.columns,        # Set column headers.
        cellLoc='center',
        loc='center',
        colWidths=[0.2] * len(temp_df.columns)
    )
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()
    plt.clf

# Low Income
create_png_dataframe(descriptive_stats_life_exp_low_income, "Descriptive_stats-Low_Income.png", "../Visualizations/Dataframes/Income_Class", (10,2))

# Lower-middle Income
create_png_dataframe(descriptive_stats_life_exp_lower_middle_income, "Descriptive_stats-Lower_middle_Income.png", "../Visualizations/Dataframes/Income_Class", (6,2))

# Upper-middle Income
create_png_dataframe(descriptive_stats_life_exp_upper_middle_income, "Descriptive_stats-Upper-middle_Income.png", "../Visualizations/Dataframes/Income_Class", (7,2))

# High Income
create_png_dataframe(descriptive_stats_life_exp_high_income, "Descriptive_stats-High_Income.png", "../Visualizations/Dataframes/Income_Class", (8,2))
```
