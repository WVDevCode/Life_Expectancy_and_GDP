#### Create datasets for top and bottom 20 gdp per capita

##### Top 20 highest grossing per capita countries dataset

```python
# Manipulating Life Expectancy df descending to retrieve top 20 gdp per capita countries data
le_top_20_gdp_2000 = life_expectancy[life_expectancy['year'] == 2000]
le_top_20_gdp_2000 = le_top_20_gdp_2000.sort_values(by='gdp_per_capita', ascending=False).reset_index(drop=True)

le_top_20_gdp_2000 = le_top_20_gdp_2000.iloc[:20]
top_20_gdp_countries_list = list(le_top_20_gdp_2000['country'])

# Sort df with top 20 countries to match template format
le_top_20_gdp = life_expectancy[life_expectancy['country'].isin(top_20_gdp_countries_list)].sort_values(by=['country', 'year'], ascending=[True, False])

# Create new column which only holds gdp values for year 2015
le_top_20_gdp['gdp_2015'] = le_top_20_gdp.apply(lambda row: row['gdp_per_capita'] if row['year'] == 2015 else pd.NA, axis=1)

# Filling years prior to 2015 with equal values for sorting
le_top_20_gdp['gdp_2015'] = le_top_20_gdp['gdp_2015'].ffill().bfill()
le_top_20_gdp = le_top_20_gdp.infer_objects(copy=False)

# Sorting dataframe to group countries together with descending gdp values
le_top_20_gdp = le_top_20_gdp.sort_values(by=['gdp_2015', 'country', 'year'], ascending=[False, True, False]).reset_index(drop=True).drop(columns='gdp_2015')

le_top_20_gdp.to_csv('../Datasets/Cleaned_datasets/EDA_Datasets/Top_Grossing_Countries-per_capita.csv')
```

##### 20 lowest grossing per capita countries dataset

```python
# Sorting Life Expectancy df ascending to retrieve bottom 20 gdp per capita countries data
le_bottom_20_gdp_2000 = life_expectancy[life_expectancy['year'] == 2000]
le_bottom_20_gdp_2000 = le_bottom_20_gdp_2000.sort_values(by='gdp_per_capita', ascending=True).reset_index(drop=True)

le_bottom_20_gdp_2000 = le_bottom_20_gdp_2000.iloc[:20]
bottom_20_gdp_countries_list = list(le_bottom_20_gdp_2000['country'])

# Sort df with bottom 20 countries to match template format
le_bottom_20_gdp = life_expectancy[life_expectancy['country'].isin(bottom_20_gdp_countries_list)].sort_values(by=['country', 'year'], ascending=[True, False])

# Create new column which only holds gdp values for year 2015
le_bottom_20_gdp['gdp_2015'] = le_bottom_20_gdp.apply(lambda row: row['gdp_per_capita'] if row['year'] == 2015 else pd.NA, axis=1)

# Filling years prior to 2015 with equal values for sorting
le_bottom_20_gdp['gdp_2015'] = le_bottom_20_gdp['gdp_2015'].ffill().bfill()
le_bottom_20_gdp = le_bottom_20_gdp.infer_objects(copy=False)

# Sorting dataframe to group countries together with descending gdp values
le_bottom_20_gdp = le_bottom_20_gdp.sort_values(by=['gdp_2015', 'country', 'year'], ascending=[False, True, False]).reset_index(drop=True).drop(columns='gdp_2015')

le_bottom_20_gdp.to_csv('../Datasets/Cleaned_datasets/EDA_Datasets/Lowest_Grossing_Countries-per_capita.csv')
```
