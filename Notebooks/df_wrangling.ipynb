# Descriptive Statistics

Create dataframe containing life expectancy mean, maximum, minimum and median values of all countries by year

## Life Expectancy

### Mean

```python
years_list_mean = []
means_list = []
for year in range(2000, 2016):
    mean = np.mean(life_expectancy['life_expectancy'].where(life_expectancy.year == year))
    years_list_mean.append(year)
    means_list.append(np.round(mean, 2))
all_means_dict = {'year': years_list_mean, 'mean': means_list}

all_means_df = pd.DataFrame(all_means_dict)
```

### Max

```python
years_list_max = []
max_list = []
max_countries = []

for year in range(2000, 2016):
    max = np.max(life_expectancy['life_expectancy'][life_expectancy.year == year])
    max_country = life_expectancy['country'][(life_expectancy.year == year) & (life_expectancy['life_expectancy'] == max)].values[0]
    years_list_max.append(year)
    max_list.append(max)
    max_countries.append(max_country)
all_max_dict = {'year': years_list_max, 'max': max_list, 'max_countries': max_countries}

all_max_df = pd.DataFrame(all_max_dict)
```

### Minimum

```python
years_list_min = []
min_list = []
min_countries = []
for year in range(2000, 2016):
    min = np.min(life_expectancy['life_expectancy'].where(life_expectancy.year == year))
    min_country = life_expectancy['country'][(life_expectancy.year == year) & (life_expectancy['life_expectancy'] == min)].values[0]
    years_list_min.append(year)
    min_list.append(min)
    min_countries.append(min_country)
all_min_dict = {'year': years_list_min, 'minimum': min_list, 'min_countries' : min_countries}

all_min_df = pd.DataFrame(all_min_dict)
```

### Median

```python
years_list_median = []
median_list = []
for year in range(2000, 2016):
    median = np.median((life_expectancy['life_expectancy'].where(life_expectancy.year == year)).dropna().reset_index(drop=True))
    years_list_median.append(year)
    median_list.append(round(median, 2))
all_median_dict = {'year': years_list_median, 'median': median_list}

all_median_df = pd.DataFrame(all_median_dict)
```

### Standard Deviation

```python
years_list_std = []
std_list = []
for year in range(2000, 2016):
    std = np.std((life_expectancy['life_expectancy'].where(life_expectancy.year == year)).dropna().reset_index(drop=True))
    years_list_std.append(year)
    std_list.append(round(std, 2))
all_std_dict = {'year': years_list_std, 'std': std_list}

all_std_df = pd.DataFrame(all_std_dict)
```

### Combining stats into single dataframe and converting it into a csv file

```python
descriptive_stats_life_exp =all_means_df.merge(all_max_df, on='year').merge(all_min_df, on='year').merge(all_median_df, on='year').merge(all_std_df, on='year')
```

```python
descriptive_stats_life_exp.to_csv('../Datasets/Cleaned_datasets/Descriptive_Statistics_Dfs/Descriptive_Stats-All_Life_expectancy.csv')
```

#### Create datasets for top and bottom 20 gdp per capita

##### Top 20 highest grossing per capita countries dataset

```python
# Manipulating Life Expectancy df descending to retrieve top 20 gdp per capita countries data
le_top_20_gdp_2000 = life_expectancy[life_expectancy['year'] == 2000]
le_top_20_gdp_2000 = le_top_20_gdp_2000.sort_values(by='gdp_per_capita', ascending=False).reset_index(drop=True)

le_top_20_gdp_2000 = le_top_20_gdp_2000.iloc[:20]
top_20_gdp_countries_list = list(le_top_20_gdp_2000['country'])

# Sort df with top 20 countries to match template format
le_top_20_gdp = life_expectancy[life_expectancy['country'].isin(top_20_gdp_countries_list)].sort_values(by=['country', 'year'], ascending=[True, False])

# Create new column which only holds gdp values for year 2015
le_top_20_gdp['gdp_2015'] = le_top_20_gdp.apply(lambda row: row['gdp_per_capita'] if row['year'] == 2015 else pd.NA, axis=1)

# Filling years prior to 2015 with equal values for sorting
le_top_20_gdp['gdp_2015'] = le_top_20_gdp['gdp_2015'].ffill().bfill()
le_top_20_gdp = le_top_20_gdp.infer_objects(copy=False)

# Sorting dataframe to group countries together with descending gdp values
le_top_20_gdp = le_top_20_gdp.sort_values(by=['gdp_2015', 'country', 'year'], ascending=[False, True, False]).reset_index(drop=True).drop(columns='gdp_2015')

le_top_20_gdp.to_csv('../Datasets/Cleaned_datasets/Top_Grossing_Countries-per_capita.csv')
```

##### 20 lowest grossing per capita countries dataset

```python
# Sorting Life Expectancy df ascending to retrieve bottom 20 gdp per capita countries data
le_bottom_20_gdp_2000 = life_expectancy[life_expectancy['year'] == 2000]
le_bottom_20_gdp_2000 = le_bottom_20_gdp_2000.sort_values(by='gdp_per_capita', ascending=True).reset_index(drop=True)

le_bottom_20_gdp_2000 = le_bottom_20_gdp_2000.iloc[:20]
bottom_20_gdp_countries_list = list(le_bottom_20_gdp_2000['country'])

# Sort df with bottom 20 countries to match template format
le_bottom_20_gdp = life_expectancy[life_expectancy['country'].isin(bottom_20_gdp_countries_list)].sort_values(by=['country', 'year'], ascending=[True, False])

# Create new column which only holds gdp values for year 2015
le_bottom_20_gdp['gdp_2015'] = le_bottom_20_gdp.apply(lambda row: row['gdp_per_capita'] if row['year'] == 2015 else pd.NA, axis=1)

# Filling years prior to 2015 with equal values for sorting
le_bottom_20_gdp['gdp_2015'] = le_bottom_20_gdp['gdp_2015'].ffill().bfill()
le_bottom_20_gdp = le_bottom_20_gdp.infer_objects(copy=False)

# Sorting dataframe to group countries together with descending gdp values
le_bottom_20_gdp = le_bottom_20_gdp.sort_values(by=['gdp_2015', 'country', 'year'], ascending=[False, True, False]).reset_index(drop=True).drop(columns='gdp_2015')

le_bottom_20_gdp.to_csv('../Datasets/Cleaned_datasets/Lowest_Grossing_Countries-per_capita.csv')
```
