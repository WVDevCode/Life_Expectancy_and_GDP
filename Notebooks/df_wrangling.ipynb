Successfully save Life_expectancy_cleaned_1.csv file to life_expectancy variable

```python
life_expectancy = pd.read_csv('../Datasets/Cleaned_datasets/Life_expectancy_cleaned_1.csv').drop(columns=['Unnamed: 0'])
```

Creating two dataframes - one for the year 2000 and one for 2015 for main comparison in 16 years

```python
life_expectancy_2000 = life_expectancy[life_expectancy.year == 2000]

life_expectancy_2015 = life_expectancy[life_expectancy.year == 2015]
```

Create dataframe of top 30 highest grossing countries per capita for 2000

```python
le_top_30_gdp_2000 = life_expectancy_2000.copy()
le_top_30_gdp_2000 = le_top_30_gdp_2000.sort_values(by='gdp_per_capita', ascending=False).reset_index(drop=True)
le_top_30_gdp_2000 = le_top_30_gdp_2000.iloc[:30]
```

Create dataframe of top 30 highest grossing countries per capita for 2015

```python
le_top_30_gdp_2015 = life_expectancy_2015.copy()
le_top_30_gdp_2015 = le_top_30_gdp_2015.sort_values(by='gdp_per_capita', ascending=False).reset_index(drop=True)
le_top_30_gdp_2015 = le_top_30_gdp_2015.iloc[:30]
```
